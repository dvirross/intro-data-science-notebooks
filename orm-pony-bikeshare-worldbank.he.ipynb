## מתארים טבלה כמחלקה

האובייקט `db` מייצג את מסד הנתונים ש־Pony עובדת איתו. השורה הראשונה בתיבה הבאה, `from pony.orm import *`, מביאה כל שם ש־Pony מציעה. `Database()` יוצרת את `db`, ו־`db.bind(...)` מחברת אותו ל־SQLite; `':memory:'` הוא שוב מסד זמני בזיכרון, כמו ב[SQL מתוך Python](https://dvirross.com/he/courses/intro-data-science/sql-from-python).

</div>

````python
from pony.orm import *

db = Database()
db.bind(provider='sqlite', filename=':memory:')
````

<div dir="rtl">

מחלקה שמייצגת טבלה נקראת **[ישות](https://dvirross.com/he/courses/intro-data-science/glossary#entity)** (entity). את שם המחלקה כותבים עם `(db.Entity)` אחריו, כמו `class Day(db.Entity):`, והסוגריים מסמנים אותה כישות של מסד הנתונים `db`: Pony שומרת את האובייקטים שלה כשורות של טבלה בשם `Day`. כל עמודה היא שורה ששמה את התכונה ואת סוגה: `Required(str)` הוא טקסט שחייב להיות לו ערך, ו־`Required(int)` הוא מספר שלם שחייב להיות לו ערך. אחרי כל הישויות, `db.generate_mapping(...)` משלימה את תיאור הטבלאות; כאשר `create_tables=True` השיטה גם יוצרת את הטבלה במסד, אם הטבלה עוד לא קיימת.

</div>

````python
class Day(db.Entity):
    date = Required(str)
    season = Required(str)
    working_day = Required(int)
    rentals = Required(int)

db.generate_mapping(create_tables=True)
print(list(db.entities))
````

## מוסיפים שורות כאובייקטים

בתוכנית, Pony מדברת עם מסד הנתונים רק בתוך בלוק `with db_session:` (ראו [db_session](https://dvirross.com/he/courses/intro-data-science/glossary#db-session)). כשהבלוק מסתיים כרגיל, Pony שומרת את השינויים; אם שגיאה עוצרת את הבלוק, הוא מבטל אותם. יצירת שורה היא יצירת אובייקט: קוראים למחלקה עם ערכי העמודות.

</div>

````python
rows = [
    ('2011-01-01', 'Winter', 0, 985),
    ('2011-01-02', 'Winter', 0, 801),
    ('2011-01-03', 'Winter', 1, 1349),
    ('2011-07-04', 'Summer', 0, 6043),
    ('2011-07-05', 'Summer', 1, 4665),
    ('2011-07-06', 'Summer', 1, 4629),
]

with db_session:
    for date, season, working_day, rentals in rows:
        Day(
            date=date, season=season, working_day=working_day,
            rentals=rentals
        )
    print(Day.select().count())
    day = Day[1]
    print(day, day.date, day.rentals)
````

<div dir="rtl">

הלולאה משתמשת ב[חילוץ](https://dvirross.com/he/courses/intro-data-science/glossary#unpacking) (unpacking) מ[לולאות ופונקציות](https://dvirross.com/he/courses/intro-data-science/python-loops-and-functions), ושש השורות הן הימים האמיתיים משיעורי ה־SQL. `Day[1]` שולפת את האובייקט שהמפתח הראשי שלו הוא 1. את ערכי העמודות שלו, שנקראים התכונות שלו, קוראים כשכותבים אחרי שם האובייקט נקודה ואחריה את שם התכונה, כמו `day.date`, בדיוק כמו ב[אובייקטים וביטויים יוצרים](https://dvirross.com/he/courses/intro-data-science/python-objects-and-generators). הדפסת האובייקט מראה את שם המחלקה ואת המפתח. השתמשו באובייקטים בתוך הבלוק `with db_session:`, שבו Pony מחוברת למסד הנתונים.

## קוראים שורות: שאילתה כביטוי יוצר

`select(...)` מקבלת ביטוי יוצר. החלק שלפני `for` אומר מה להחזיר, החלק שאחרי `in` אומר מאיזו ישות, והתנאי אחרי `if` מסנן את השורות, בדיוק כמו `SELECT`, `FROM` ו־`WHERE`. מעבר בלולאה על התוצאה מריץ את השאילתה ונותן אובייקט אחד לכל שורה. השיטה `order_by` ממיינת את שורות התוצאה לפי הערך של הפונקציה הקטנה שנותנים לה, ו־`desc(...)` הופכת את הסדר לסדר מהגדול אל הקטן, כמו `ORDER BY ... DESC`.

</div>

````python
with db_session:
    summer = select(d for d in Day if d.season == 'Summer')
    for d in summer.order_by(lambda d: desc(d.rentals)):
        print(d.date, d.rentals)
````

<div dir="rtl">

בישות קוראים עמודה לפי שם, כמו `d.rentals`, ולכן סדר העמודות אינו משנה, ושם שגוי כמו `d.rentalz` עוצר כבר בשורה של השאילתה, עם הודעה שמזכירה אותו.

כדי לראות את ה־SQL ש־Pony כתבה, השיטה `get_sql` של שאילתה מחזירה אותו כטקסט:

</div>

````python
with db_session:
    print(
        select(
            d for d in Day if d.season == 'Summer'
        ).get_sql()
    )
````

<div dir="rtl">

זו שאילתה שהייתם יכולים להקליד ב־DB Browser. Pony משתמשת גם ב[**פרמטרים**](https://dvirross.com/he/courses/intro-data-science/glossary#argument) מ[SQL מתוך Python](https://dvirross.com/he/courses/intro-data-science/sql-from-python): משתנה של Python בתוך הביטוי אינו מודבק לטקסט, אלא נשלח כערך עבור סימן שאלה.

</div>

````python
with db_session:
    limit = 4600
    busy = select(d for d in Day if d.rentals > limit)
    print(busy.get_sql())
    print([d.date for d in busy])
````

<div dir="rtl">

**[חיתוך](https://dvirross.com/he/courses/intro-data-science/glossary#slice)** (slice) לוקח רק את השורות הראשונות, כמו ברשימה: `select(...)[:2]` מחזירה את שני האובייקטים הראשונים.

כשצריך רק אובייקט אחד, השיטה `first()` קצרה מ־`list(...)[0]`. היא מחזירה את האובייקט הראשון בתוצאה, ו־`None` כשהתוצאה ריקה, ואילו `list(...)[0]` עוצרת בשגיאה על רשימה ריקה.

`first()` שונה גם מהחיתוך `[:1]`: החיתוך מחזיר רשימה שיש בה לכל היותר אובייקט אחד, ו־`first()` מחזירה את האובייקט עצמו. אחרי `order_by(...)` האובייקט הראשון הוא הראשון לפי אותו סדר. התיבה לוקחת את היום עם הכי מעט השכרות, ואחר כך מבקשת יום עם יותר מ־10000 השכרות, שאינו קיים:

</div>

````python
with db_session:
    quietest = select(d for d in Day).order_by(
        lambda d: d.rentals
    ).first()
    print(quietest.date, quietest.rentals)
    huge = select(d for d in Day if d.rentals > 10000).first()
    print(huge)
````

## משנים ומוחקים שורות

כדי לשנות עמודה, מציבים ערך בתכונה של האובייקט. כדי למחוק שורה, קוראים ל־`delete()` של האובייקט שלה. קודם נוסיף **יום בדיקה** שאינו נתון אמיתי, כדי שהימים האמיתיים יישארו כפי שהם. `Day.get(...)` מחזירה את האובייקט היחיד שמתאים, או `None` אם אין כזה. **שערו:** הבלוק השני משנה את `rentals` ל־2, והשלישי הוא `with db_session:` חדש. מה ידפיס השלישי בשורה הראשונה שלו: 1, 2 או שגיאה?

</div>

````python
with db_session:
    Day(
        date='2099-01-01', season='Summer', working_day=1,
        rentals=1
    )

with db_session:
    test_day = Day.get(date='2099-01-01')
    test_day.rentals = 2
    print(test_day.rentals)

with db_session:
    print(Day.get(date='2099-01-01').rentals)
    Day.get(date='2099-01-01').delete()
    print(count(d for d in Day))
    print(Day.get(date='2099-01-01'))
````

<div dir="rtl">

השוו ל[SQL מתוך Python](https://dvirross.com/he/courses/intro-data-science/sql-from-python), שבו כתבתם `UPDATE` ו־`DELETE` כטקסט SQL והחלטתם מתי לקרוא ל־`commit()`. כאן סוף בלוק ה־`db_session` הוא ה־commit, ושגיאה בתוך הבלוק היא ה־rollback. אפשר גם לבטל בכוונה: הקריאה `rollback()` בתוך הבלוק מבטלת כל שינוי שנעשה מאז תחילת הבלוק, כמו ש־`conn.rollback()` עושה ב[סעיף על שינוי נתונים ושמירתם](https://dvirross.com/he/courses/intro-data-science/sql-from-python#שינוי-נתונים-ושמירתם), והבלוק ממשיך עם מסד הנתונים כפי שהיה.

הקריאה `commit()` בתוך הבלוק שומרת מיד את השינויים שנעשו עד כה, והבלוק ממשיך. `rollback()` שבאה אחריה מבטלת רק את מה שבא אחרי ה־`commit()`. התיבה שלהלן מוסיפה יום בדיקה, שומרת אותו עם `commit()`, מוסיפה יום בדיקה שני ומבטלת אותו עם `rollback()`. אחר כך בלוק חדש קורא מה מסד הנתונים מחזיק, ומוחק את יום הבדיקה הראשון, כדי שבטבלה יהיו שוב ששת הימים האמיתיים:

</div>

````python
with db_session:
    Day(
        date='2097-01-01', season='Summer', working_day=1,
        rentals=1
    )
    commit()
    Day(
        date='2097-01-02', season='Summer', working_day=1,
        rentals=2
    )
    rollback()

with db_session:
    print(Day.get(date='2097-01-01').rentals)
    print(Day.get(date='2097-01-02'))
    Day.get(date='2097-01-01').delete()
    print(count(d for d in Day))
````

<div dir="rtl">

יום הבדיקה הראשון נשמר בידי `commit()`, ולכן הוא עדיין שם, והשני בוטל לפני שנשמר, ולכן `Day.get` לא מוצאת דבר ומחזירה `None`.

## האם מדינות עשירות חיות יותר?

שלוש אזהרות מלוות את הנתונים. מעט מדינות אינן העולם, ולכן המבחר יכול להראות דפוס אבל לא למדוד אותו. שתי מדידות שנעות יחד הן קשר סטטיסטי ולא הוכחה שאחת גורמת לשנייה. בנוסף, ערך שאין למקור נשאר חסר: הוא אינו אפס. התוצר של סוריה ב־2000 חסר במקור, ולכן הוא חסר גם כאן.

השורה הראשונה בתיבה הבאה מביאה את השמות של Pony, ו־`geo` הוא אובייקט מסד הנתונים של השיעור הזה; `':memory:'` הוא מסד נתונים זמני בזיכרון, כמו קודם. כמה שמות מתארים את הקשר ואת סוגי העמודות החדשים:

- `Set('Observation')` היא התכונה `observations` של מדינה. היא מחזיקה את הקבוצה של כל המדידות שמצביעות על המדינה הזו, והיא אינה עמודה בטבלה: Pony קוראת אותה מהעמודה שבצד השני. שם המחלקה בגרשיים כי `Observation` מוגדרת מתחתיה.
- `Required(Country)` היא התכונה `country` של מדידה. ערכה חייב להיות אובייקט `Country` אחד, ולכן כל מדידה שייכת למדינה. `Optional(Country)` היתה מאפשרת מדידה בלי מדינה.
- `Optional(float)` היא מספר עם נקודה עשרונית שיכול להיות חסר, ו־`Optional(int)` היא מספר שלם שיכול להיות חסר. ערך חסר הוא `None` ב־Python ו־`NULL` ב־SQL.

הפונקציה `show(Observation)` מדפיסה את התיאור של ישות כפי ש־Pony הבינה אותו: התכונות שלה והסוגים שלהן, כולל ה־`id` ש־Pony הוסיפה.

</div>

````python
from pony.orm import *

geo = Database()
geo.bind(provider='sqlite', filename=':memory:')

class Country(geo.Entity):
    code = Required(str)
    name = Required(str)
    income_level = Required(str)
    observations = Set('Observation')

class Observation(geo.Entity):
    year = Required(int)
    life_expectancy = Optional(float)
    gdp_per_capita = Optional(float)
    population = Optional(int)
    country = Required(Country)

geo.generate_mapping(create_tables=True)
show(Observation)
````

### יוצרים את האובייקטים ומחברים ביניהם

המבחר הוא שתי רשימות של טאפלים: המדינות, והמדידות, וכל מדידה עם קוד המדינה שלה, השנה, תוחלת החיים, התוצר לנפש והאוכלוסייה. הערכים מעוגלים (שלוש ספרות אחרי הנקודה העשרונית לתוחלת החיים, שתיים לתוצר לנפש), והעיגול אינו משנה אף תוצאה בשיעור.

</div>

````python
country_rows = [
    ('ISR', 'Israel', 'High income'),
    ('JPN', 'Japan', 'High income'),
    ('USA', 'United States', 'High income'),
    ('NOR', 'Norway', 'High income'),
    ('BRA', 'Brazil', 'Upper middle income'),
    ('CHN', 'China', 'Upper middle income'),
    ('EGY', 'Egypt, Arab Rep.', 'Lower middle income'),
    ('IND', 'India', 'Lower middle income'),
    ('ETH', 'Ethiopia', 'Low income'),
    ('SYR', 'Syrian Arab Republic', 'Low income'),
]
print(len(country_rows))
````

````python
observation_rows = [
    ('ISR', 2000, 78.954, 25746.83, 6289000),
    ('ISR', 2020, 82.649, 40969.3, 9215100),
    ('JPN', 2000, 81.076, 27694.54, 126843000),
    ('JPN', 2020, 84.56, 43560.02, 126261000),
    ('USA', 2000, 76.637, 36329.97, 282162411),
    ('USA', 2020, 76.98, 64465.3, 331578104),
    ('NOR', 2000, 78.634, 36813.84, 4490967),
    ('NOR', 2020, 83.21, 69779.29, 5379475),
    ('BRA', 2000, 69.584, 9091.77, 174018282),
    ('BRA', 2020, 74.506, 16101.62, 208660842),
    ('CHN', 2000, 72.288, 2963.43, 1262645000),
    ('CHN', 2020, 78.019, 18267.48, 1411100000),
    ('EGY', 2000, 67.332, 5451.52, 73083284),
    ('EGY', 2020, 69.79, 15231.87, 109315124),
    ('IND', 2000, 62.749, 2030.77, 1057922733),
    ('IND', 2020, 70.156, 6766.06, 1402617695),
    ('ETH', 2000, 50.88, 473.9, 67411494),
    ('ETH', 2020, 65.969, 2406.79, 118917671),
    ('SYR', 2000, 70.856, None, 16637888),
    ('SYR', 2020, 71.944, 3878.43, 21049429),
]
print(len(observation_rows))
````

<div dir="rtl">

ערכים במילון יכולים להיות אובייקטים כלשהם, לא רק מספרים וטקסט. נבנה מילון שהמפתחות שלו הם קודי המדינות והערכים שלו הם אובייקטי `Country`. כדי ליצור מדידה של מדינה, נחפש את אובייקט המדינה לפי הקוד שלה ונעביר אותו כ־`country=...`. כך מחברים בין שתי השורות: Pony שומרת את המפתח של המדינה בעמודה `country` של המדידה. הערך `None` בשורה נכנס לעמודה `Optional` כערך חסר.

</div>

````python
with db_session:
    countries = {}
    for code, name, income in country_rows:
        countries[code] = Country(
            code=code, name=name, income_level=income
        )
    for code, year, life, gdp, people in observation_rows:
        Observation(
            year=year, life_expectancy=life,
            gdp_per_capita=gdp, population=people,
            country=countries[code]
        )
    print(
        Country.select().count(),
        Observation.select().count()
    )
````

### עוקבים אחרי הקשר בשני הכיוונים

בתוך `db_session` נקודה עוקבת אחרי הקשר. עבור מדידה `obs`, הביטוי `obs.country` הוא אובייקט ה־`Country` שלה, ו־`obs.country.name` הוא שם המדינה. בכיוון ההפוך, `country.observations` היא קבוצת המדידות של מדינה. אפשר לעבור על קבוצה בלולאה כמו על רשימה, ויש לה שלוש שיטות: `count()` מחזירה את מספר האובייקטים בה, `order_by(...)` ממיינת אותה כמו בשאילתה, ו־`select(...)` משאירה את האובייקטים שהפונקציה שנותנים לה מחזירה עבורם אמת. כדי לקחת אובייקט אחד לפי מקומו, כמו מרשימה, הופכים קודם את התוצאה של `order_by` או של `select` לרשימה עם `list(...)`.

</div>

````python
with db_session:
    israel = Country.get(code='ISR')
    print(israel.name, israel.income_level)
    print(israel.observations.count())
    in_2020 = list(
        israel.observations.select(lambda o: o.year == 2020)
    )
    obs = in_2020[0]
    print(
        obs.country.name, obs.year,
        round(obs.life_expectancy, 1)
    )
````

<div dir="rtl">

הפונקציה `round(x, 1)` מעגלת את `x` לספרה אחת אחרי הנקודה העשרונית.

השיטה `first()` של שאילתה עובדת גם על קבוצה של אובייקטים קשורים. התיבה לוקחת את המדידה האחרונה של ישראל, ואחר כך מחפשת שנה שאין לה מדידה:

</div>

````python
with db_session:
    israel = Country.get(code='ISR')
    latest = israel.observations.order_by(
        lambda o: desc(o.year)
    ).first()
    print(latest.year, round(latest.life_expectancy, 1))
    missing = israel.observations.select(
        lambda o: o.year == 1990
    ).first()
    print(missing)
````

### שאלות שהאובייקטים הקשורים עונים עליהן

לכל מדינה, `country.observations.order_by(lambda o: o.year)` מחזיקה את שתי המדידות שלה, השנה המוקדמת קודם, ו־`list(...)` הופכת אותה לרשימה, כך שהמקום `[0]` הוא 2000 והמקום `[1]` הוא 2020. השינוי בתוחלת החיים הוא הערך של 2020 פחות הערך של 2000. נאסוף כל שינוי עם שם המדינה שלו ברשימה של טאפלים, ו־`sorted(..., reverse=True)` מציבה את השינוי הגדול ראשון:

</div>

````python
with db_session:
    gains = []
    for country in Country.select():
        years = list(
            country.observations.order_by(lambda o: o.year)
        )
        early = years[0].life_expectancy
        late = years[1].life_expectancy
        gains.append((round(late - early, 1), country.name))
    ranked = sorted(gains, reverse=True)
    print(ranked[:3])
    print(ranked[-1])
````

<div dir="rtl">

תוחלת החיים עלתה בכל מדינה במבחר, והכי הרבה באתיופיה, שהתחילה מהנמוכה ביותר. המבחר אינו אומר מדוע. שתי השנים הן תצלום מצב, והשינוי הקטן ביותר, 0.3 שנים בארצות הברית, יכול לנבוע מהמגמה הארוכה, מכך ש־2020 היא שנת המגפה, או משניהם; שתי נקודות אינן יכולות להפריד ביניהם.

בתוצר לנפש ערך יכול להיות חסר, ואז אין מה לחלק. הבדיקה `x is None` מתקיימת כש־`x` הוא הערך החסר, ולכן הלולאה מפרידה את המדינות שאין להן ערך, ולשאר היא מחשבת פי כמה התוצר גדל: הערך של 2020 חלקי הערך של 2000.

</div>

````python
with db_session:
    multiples = []
    missing = []
    for country in Country.select():
        years = list(
            country.observations.order_by(lambda o: o.year)
        )
        early = years[0].gdp_per_capita
        late = years[1].gdp_per_capita
        if early is None or late is None:
            missing.append(country.name)
        else:
            multiples.append(
                (round(late / early, 1), country.name)
            )
    print(sorted(multiples, reverse=True)[:3])
    print('missing:', missing)
````

<div dir="rtl">

סוריה אינה בדירוג כי הערך שלה ל־2000 חסר, ולא כי הוא אפס. ההפרש עונה על שאלה אחרת מהכפולה: באתיופיה, שהכפולה שלה 5.1, התוצר לנפש גדל ב־1,933 דולר (2,407 פחות 474), ובנורווגיה, שאינה בין שלוש הכפולות הגדולות, ב־32,965 (69,779 פחות 36,814).

### תנאי שחוצה את הקשר

בתוך שאילתה נקודה יכולה לעקוב אחרי הקשר גם כן: `o.country.income_level` היא רמת ההכנסה של המדינה של המדידה `o`. שאילתה כזו צריכה את שתי הטבלאות, ו־Pony כותבת את החיבור שמדריך ה־SQL כתב ביד. נקבץ את המדידות של 2020 לפי רמת הכנסה ונחשב ממוצע של שני המדדים; `avg` מדלגת על ערכים חסרים:

</div>

````python
with db_session:
    by_level = select(
        (
            o.country.income_level,
            avg(o.life_expectancy),
            avg(o.gdp_per_capita)
        )
        for o in Observation if o.year == 2020
    )
    print(by_level.get_sql())
    by_gdp = sorted(by_level, key=lambda row: row[2])
    for level, life, gdp in by_gdp:
        print(level, round(life, 1), round(gdp))
````

<div dir="rtl">

ב־`WHERE` יש את התנאי על השנה ואת השורה שמחברת את המפתח הזר `"o"."country"` למפתח הראשי `"country"."id"`, כמו ש־`ON` עושה ב־`JOIN`. לשאלה שפתחנו בה אפשר לענות כך: במבחר הזה רמות הכנסה גבוהות יותר מלוות תוחלת חיים ארוכה יותר, וזה קשר סטטיסטי בין ממוצעים של שתיים עד ארבע מדינות בכל קבוצה. הוא אינו מראה שהכנסה מייצרת בריאות, או להפך.

טבלה מגיעה לא פעם כ[קובץ CSV](https://dvirross.com/he/courses/intro-data-science/python-values-and-collections#סוגי-קבצי-נתונים): קובץ טקסט עם שורה אחת בכל שורה ועם שמות העמודות בשורה הראשונה. הקובץ `world-bank-more-countries.csv` מכיל את אותם שלושה מדדים, ערכים מלאים בלי עיגול, לארבע מדינות נוספות, לשנים 2000 ו־2020, עם שורה לכל מדינה ושנה. המקור והרישיון הם של התמצית שלמעלה; שני המדדים הראשונים בקובץ נשלפו ב־5 בנובמבר 2026 והאוכלוסייה ב־8 בנובמבר 2026. התוצר לנפש של אריתריאה ב־2020 חסר במקור, ולכן השדה הזה בקובץ ריק.

Python קוראת קובץ כזה עם המודול `csv`. הפונקציה `open(name, newline='', encoding='utf-8')` פותחת קובץ, ובלוק `with` סוגר אותו כשהבלוק מסתיים; האפשרות `newline=''` היא מה שהמודול `csv` מבקש. הפונקציה `csv.DictReader(file)` עוברת על השורות שאחרי הראשונה, ונותנת כל שורה כ[מילון](https://dvirross.com/he/courses/intro-data-science/glossary#dictionary) שהמפתחות שלו הם שמות העמודות. כל ערך שמגיע מקובץ הוא טקסט, גם מספר, ושדה ריק הוא הטקסט הריק `''`. הפונקציה `int(text)` הופכת טקסט למספר שלם, ו־`float(text)` הופכת טקסט למספר עם נקודה עשרונית.

</div>

````python
import csv

with open(
    'world-bank-more-countries.csv', newline='', encoding='utf-8'
) as file:
    rows = list(csv.DictReader(file))

print(len(rows))
print(rows[0]['name'], rows[0]['year'] * 2)
print(rows[0]['name'], int(rows[0]['year']) + 1)
print(rows[7]['gdp_per_capita'] == '')
````

טעינת השורות לישויות היא הלולאה של [יוצרים את האובייקטים ומחברים ביניהם](#יוצרים-את-האובייקטים-ומחברים-ביניהם), עם `row['code']` במקום הקוד המחולץ. `Country` נוצרת רק בפעם הראשונה שהקוד שלה מופיע: הבדיקה `code not in countries` מתקיימת כשבמילון אין עדיין מפתח `code`. השיטה `get` מקבלת כמה תנאים בבת אחת: `Observation.get(year=2020, country=some_country)` מחזירה את המדידה האחת שמתאימה לכולם.

## לזכור

- **קשר של אחד לרבים** הוא מפתח זר שנראה משני הצדדים: `Required(Country)` או `Optional(Country)` בצד של הרבים, ו־`Set('Observation')` בצד של האחד. נקודה עוקבת אחריו בשני הכיוונים, ובתוך שאילתה נקודה כמו `o.country.income_level` גורמת ל־Pony לכתוב את החיבור.
- `Optional(float)` מאפשרת ערך חסר, שהוא `None`; בודקים אותו עם `is None` לפני שמחשבים. ערך חסר אינו אפס.
- מילון מקוד לאובייקט (`countries[code]`) מחבר אובייקטים חדשים לאובייקט הקשור הנכון.
- `csv.DictReader` קוראת קובץ CSV: כל ערך הוא טקסט, ולכן ממירים אותו עם `int(...)` או `float(...)`, ושדה ריק הופך ל־`None`.
- `show(Entity)` מדפיסה איך Pony הבינה ישות, ו־`set_sql_debug(True)` מדפיסה כל פקודת SQL שרצה. כדי להשתמש בקובץ מסד נתונים שקיים, מחברים אותו עם `bind` בלי `create_db=True` וקוראים ל־`generate_mapping()` בלי `create_tables=True`.
