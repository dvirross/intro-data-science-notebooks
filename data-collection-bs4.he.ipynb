התא הבא מגדיר את הפונקציה `check(task_number)`. אחרי שמריצים את התא של משימה, מריצים את התא שמתחתיו: `check` מריצה שוב את התא שלכם, אוספת את מה שהוא מדפיס ומשווה אותו לפלט הצפוי. היא אומרת אם הפלט נכון, ואם לא באיזו שורה הוא שונה; היא לא מראה את הפתרון.

</div>

````python
import ast, contextlib, hashlib, io

# The expected output of each task, as hashes of its lines (the solutions are not stored here).
EXPECTED = {int(k): v for k, v in {
    '1': [
        'a4ca3edcead4'
    ],
    '2': [
        '2dfe249ea3aa'
    ],
    '3': [
        'c9ad4f15cacd'
    ],
    '4': [
        'cacbc3d3f1c4'
    ],
    '5': [
        '19581e27de7c'
    ],
    '6': [
        '7b1bd53949b7'
    ],
    '7': [
        '4e07408562be'
    ],
    '8': [
        'ead42beead86'
    ],
    '9': [
        '4652614c4d87'
    ]
}.items()}


def check(task):
    """Run the last cell that ran (not a check) again and compare what it prints with the expected output of the task."""
    if task not in EXPECTED:
        print('אין משימה כזו: ', task)
        return
    cells = [s for s in In[:-1] if s.strip() and not s.lstrip().startswith('check(')]   # In: the cells that ran, kept by IPython
    tree = ast.parse(cells[-1] if cells else '')
    last = tree.body.pop() if tree.body and isinstance(tree.body[-1], ast.Expr) else None   # a bare last expression counts as printed
    out = io.StringIO()
    try:
        with contextlib.redirect_stdout(out):
            exec(compile(tree, '<task>', 'exec'), globals())
            if last is not None:
                value = eval(compile(ast.Expression(last.value), '<task>', 'eval'), globals())
                if value is not None:
                    print(repr(value))
    except Exception as error:
        print('התא נכשל בשגיאה:', type(error).__name__ + ':', error)
        return
    got = [hashlib.sha256(line.rstrip().encode()).hexdigest()[:12] for line in out.getvalue().rstrip().splitlines()]
    want = EXPECTED[task]
    if not got:
        print('לא הודפס כלום. הדפיסו את התשובה בתא המשימה, הריצו אותו, והריצו את `check` מיד אחריו.')
    elif len(got) != len(want):
        print('התא הדפיס {got} שורות והמשימה מצפה ל־{want}. קראו שוב את המשימה ואת הרמז.'.format(got=len(got), want=len(want)))
    elif got != want:
        print('שורה {k} שונה מהפלט הצפוי. קראו שוב את המשימה ואת הרמז.'.format(k=next(i for i in range(len(want)) if got[i] != want[i]) + 1))
    else:
        print('OK: הפלט כצפוי.')
````

## דף הוא טקסט עם תגיות

כשמקלידים כתובת בדפדפן, הדפדפן שולח **בקשה** (request) אל ה**שרת** שמחזיק את הדף, והשרת שולח בחזרה **תשובה** (response). ההחלפה הזו נעשית לפי הכללים של [HTTP](https://dvirross.com/he/courses/intro-data-science/glossary#http), הדרך שבה האינטרנט מדבר, והכתובת היא [URL](https://dvirross.com/he/courses/intro-data-science/glossary#url). התשובה מחזיקה את הטקסט של הדף, כתוב ב־[HTML](https://dvirross.com/he/courses/intro-data-science/glossary#html): טקסט שעטוף ב**תגיות** (tags), מילים בסוגריים זוויתיים כמו `<dt>` ו־`</dt>`, שאומרות לדפדפן מה כל חלק: כותרת, קישור, רשימה. הדפדפן מסתיר את התגיות ומצייר את הדף. תוכנית שמבקשת את הדף מקבלת את התגיות גם כן.

הדף בנוי מ**אלמנטים** (elements). אלמנט הוא חלק של הדף שבין תגית פותחת לתגית סוגרת, והוא יכול להחזיק טקסט ואלמנטים אחרים. הנה תחילת הרשומה הראשונה ברשימה, כפי שהיא בטקסט של הדף, עם חלקים שהושמטו:

```html
<dt><a name='item1'>[1]</a>
  <a href="/abs/2610.00106" title="Abstract">arXiv:2610.00106</a>
<dd>
  <div class='meta'>
    <div class='list-title mathjax'><span class='descriptor'>Title:</span>
      SyntheticHLS: Building Diverse Synthetic High-Level ...
    </div>
    <div class='list-authors'>
      <a href="https://arxiv.org/search/...">Stefan Abi-Karam</a>, ...
    </div>
```

- **קינון.** האלמנט `dd` מחזיק את ה־`div` מהסוג `meta`, שמחזיק את ה־`div` של הכותרת, שמחזיק `span`. האלמנטים הפנימיים הם ה**ילדים** (children) של החיצוני.
- **תכונות.** תגית פותחת יכולה לשאת תכונות (attributes), זוגות של שם וערך כמו `href="/abs/2610.00106"` או `title="Abstract"`. הטקסט של אלמנט הוא מה שהקורא רואה; תכונה היא מידע על האלמנט, למשל לאן קישור מוביל.
- **סוג.** התכונה `class` נותנת לאלמנט שם אחד או יותר, כאן `list-title` ו־`mathjax`. אלמנטים מאותו סוג חולקים שם, ולכן ה־`class` הוא הדרך של תוכנית להבדיל בין כותרת לשורת מחברים.

הדפדפן מראה את האלמנטים של הדף כעץ. לחיצה ימנית על חלק של הדף ובחירה ב**בדיקה** (Inspect) פותחות את כלי המפתחים, והאלמנט שלחצנו עליו מסומן. כשרוצים נתונים מדף, המבט הזה מחליט מה לכתוב בקוד: באיזו תגית הם נמצאים, איזה `class` או איזו תכונה מבדילים אותם מהאלמנטים שסביבם, והאם צריך את הטקסט שלהם או תכונה.

הספרייה `requests` שולחת בקשה מ־Python. השורה הראשונה בתיבה הבאה מביאה אותה. הפונקציה `requests.get` שולחת בקשה שמבקשת דף, ומחזירה **אובייקט תשובה** (response object): כל מה שחזר, שמור במשתנה אחד. שני דברים בו נקראים קודם: `status_code`, מספר שאומר איך הבקשה עברה (200 אומר שהדף הגיע), ו־`text`, הטקסט של הדף כמחרוזת אחת ארוכה. הפונקציה `len` מחזירה את מספר התווים של מחרוזת. הכתובת ארוכה, ולכן היא כתובה בשני חלקים: שתי מחרוזות צמודות במירכאות, בתוך סוגריים, מחוברות ב־Python למחרוזת אחת, והסוגריים מאפשרים להן להמשיך בשורה הבאה.

## Beautiful Soup קוראת את התגיות

**Beautiful Soup** היא ספרייה של Python שקוראת את הטקסט של דף HTML ובונה ממנו את העץ של האלמנטים שלו, כדי שנוכל לבקש אלמנט במקום לחתוך טקסט. תוכנית שקוראת טקסט בפורמט נתון ובונה ממנו מבנה היא [**מנתח**](https://dvirross.com/he/courses/intro-data-science/glossary#parser) (parser). הספרייה `beautifulsoup4` מיובאת בשם `bs4`, והמחלקה שבה משתמשים היא `BeautifulSoup`. היא מקבלת שני ארגומנטים: את הטקסט של הדף, ואת השם של המנתח, `'html.parser'`, זה שמגיע עם Python.

צריך עוד שני שמות. `soup.title` הוא האלמנט הראשון מהתגית `title` בדף, והשם `text` שנכתב אחרי אלמנט, כמו ב־`soup.title.text`, נותן את כל הטקסט שבתוכו.

</div>

````python
from bs4 import BeautifulSoup

soup = BeautifulSoup(html, 'html.parser')
print(soup.title)
print(soup.title.text)
````

<div dir="rtl">

השורה הראשונה הדפיסה את האלמנט עם התגיות שלו, והשנייה רק את הטקסט שהוא מחזיק.

## `find`: האלמנט הראשון שמתאים

השיטה `find` של ה־`soup` מחזירה את האלמנט הראשון שמתאים למה שביקשנו. נותנים לה שם של תגית, ובנוסף, כדי להבדיל בין אלמנטים מאותה תגית, `class`. המילה `class` שמורה ב־Python, ולכן הארגומנט נכתב `class_`.

</div>

````python
title = soup.find('div', class_='list-title')
print(title.text)
````

<div dir="rtl">

הטקסט של האלמנט הוא מה שהכותרת מחזיקה, עם האלמנט הפנימי שלה: המילה `Title:` של ה־`span`, והרווחים ומעברי השורה של הדף סביב הכותרת. שתי שיטות של מחרוזת מנקות אותו. `replace('Title:', '')` מחליפה חלק מהטקסט בטקסט אחר, כאן בכלום, ו־`strip()` מסירה את הרווחים ואת מעברי השורה משני הקצוות של המחרוזת.

</div>

````python
print(title.text.replace('Title:', '').strip())
````

## תכונה נקראת עם `get`

הקישור לדף של המאמר הראשון אינו טקסט על המסך: הוא התכונה `href` של אלמנט `a`. השיטה `get` של אלמנט מחזירה את הערך של תכונה אחת, לפי שמה. כדי למצוא את האלמנט, `find` מקבלת כל תכונה כארגומנט: האלמנט `a` שהתכונה `title` שלו היא `Abstract`.

</div>

````python
link = soup.find('a', title='Abstract')
print(link.text.strip())
print(link.get('href'))
````

הארגומנט `attrs` כותב תנאי על תכונה כמילון. הוא נחוץ כשתכונה נקראת כמו פרמטר של `find`: הפרמטר הראשון הוא `name`, שם התגית, ולכן `find('a', name='item1')` זורקת `TypeError`. את העוגן `item1` מוצאים עם `attrs`:

## לאיזה מאמר יש הכי הרבה מחברים?

מאמרים בארכיטקטורת חומרה הם לעיתים קרובות עבודה של צוות: לחלקם מחבר אחד ולחלקם עשרה. ברשימה של arXiv לאוקטובר 2026 אנחנו שואלים כמה מחברים יש למאמר, לאיזה מאמר יש הכי הרבה, וכמה יש בממוצע. העותק השמור של הרשימה מחזיק את 25 הרשומות הראשונות מתוך 266, ולכן התשובות מתארות את 25 המאמרים האלה ולא את החודש כולו.

התיבה הראשונה מבקשת את העותק השמור של הרשימה ובונה את ה־`soup`, כמו ב[מדף אינטרנט לנתונים: Beautiful Soup](https://dvirross.com/he/courses/intro-data-science/data-collection-beautifulsoup), כדי שהדף הזה יעמוד בפני עצמו.

</div>

````python
import requests
from bs4 import BeautifulSoup

list_url = (
    'https://dvirross.com/materials/courses/'
    'intro-data-science/arxiv-cs-ar-2026-10.htm'
)
page_text = requests.get(list_url).text
soup = BeautifulSoup(page_text, 'html.parser')
print(soup.title.text)
````

<div dir="rtl">

## `find_all`: כל האלמנטים שמתאימים

כל מאמר ברשימה הוא אלמנט `dd` שמחזיק את הכותרת, את המחברים ואת הנושאים שלו. השיטה `find_all` מחזירה את כל האלמנטים שמתאימים, כרשימה, בסדר של הדף. היא מקבלת את אותם ארגומנטים כמו `find`.

</div>

````python
entries = soup.find_all('dd')
print(len(entries))
````

<div dir="rtl">

ברשימה `entries` יש 25 אלמנטים, אחד לכל מאמר, ולכל אחד מהם `find` משלו שמחפשת רק בתוכו. הצעדים שקוראים את הכותרת של מאמר אחד נכנסים לפונקציה, `title_of`. הבנת רשימה בונה רשימה מרשימה אחרת בשורה אחת: הביטוי שלפני `for` מחושב לכל אלמנט, כאן הקריאה ל־`title_of`.

</div>

````python
def title_of(entry):
    box = entry.find('div', class_='list-title')
    return box.text.replace('Title:', '').strip()

titles = [title_of(entry) for entry in entries]
print(len(titles))
print(titles[0])
print(titles[1])
````

<div dir="rtl">

## סופרים את המחברים

המחברים של מאמר נמצאים ב־`div` מהסוג `list-authors`, וכל מחבר הוא קישור, אלמנט `a`. לכן מספר המחברים הוא מספר הקישורים בתוך ה־`div` הזה: `find_all` מחזירה רשימה ו־`len` סופרת אותה. הריצו את התיבה והסתכלו על התוצאה למאמר השני, `entries[1]`: מספר הקישורים ב־`list-authors` שלו.

</div>

````python
authors = entries[1].find('div', class_='list-authors')
print(len(authors.find_all('a')))
````

<div dir="rtl">

הצעדים שקוראים מאמר אחד זהים לכל המאמרים, ולכן שמים אותם בפונקציה, שמחזירה את הזוג של הכותרת ומספר המחברים, וקוראים לה על כל רשומה. זוג כזה הוא טאפל.

</div>

````python
def read_entry(entry):
    authors = entry.find('div', class_='list-authors')
    return (title_of(entry), len(authors.find_all('a')))

papers = [read_entry(entry) for entry in entries]
print(papers[0])
````

<div dir="rtl">

## המקסימום והממוצע

`papers` היא רשימה של 25 זוגות. הפונקציה `max` מחזירה את הפריט הגדול ביותר, והארגומנט `key` שלה אומר במה להשוות: כאן בפונקציה שמקבלת זוג ומחזירה את החלק השני שלו, מספר המחברים, שנכתבת `lambda paper: paper[1]`. הממוצע הוא הסכום חלקי הכמות, ו־`round(number, 2)` משאירה שתי ספרות אחרי הנקודה העשרונית.

</div>

````python
top = max(papers, key=lambda paper: paper[1])
print(top[1], top[0])

counts = [paper[1] for paper in papers]
print(round(sum(counts) / len(counts), 2))
````

<div dir="rtl">

## טקסט שמחזיק מספר

מספר עם נקודה עשרונית ממירים עם `float`, ויחידה שכתובה אחריו מסירים קודם. הדף של מאמר כותב את הגודל של ההגשה שלו כ־`3,330 KB`, ואותם צעדים נותנים את המספר:

## אלמנט שחסר

לא לכל מאמר יש שורת הערות, שבה המחברים כותבים, למשל, את מספר העמודים. ברשימה ל־13 מ־25 המאמרים אין בכלל `div` מהסוג `list-comments`. לרשומה כזו `find` מחזירה `None`, ו־`None.text` היא שגיאה. תוכנית שקוראת הרבה רשומות בודקת אם יש `None` לפני שהיא קוראת.

</div>

````python
without_comments = [
    entry for entry in entries
    if entry.find('div', class_='list-comments') is None
]
print(len(without_comments))
````

<div dir="rtl">

כשההערות קיימות, הן טקסט חופשי שהמחברים כתבו, כמו `7 pages, 5 figures. Accepted to ASP-DAC 2027`. תוכנית קוראת מספר עמודים רק במקום שהטקסט אומר אותו, בצורה שמספר בא קודם והמילה `pages` אחריו. השיטה `isdigit()` אומרת אם טקסט מורכב מספרות בלבד, ו־`startswith('pages')` אם טקסט מתחיל במילה. הפונקציה מחזירה `None` לכל רשומה שלא אומרת.

</div>

````python
def pages_of(entry):
    box = entry.find('div', class_='list-comments')
    if box is None:
        return None
    words = box.text.replace('Comments:', '').split()
    if len(words) < 2:
        return None
    if words[0].isdigit() and words[1].startswith('pages'):
        return int(words[0])
    return None
````

````python
all_pages = [pages_of(entry) for entry in entries]
print(all_pages)
````

מספר חסר אינו אפס. כדי לחשב עם המספרים שקיימים, שומרים רק את הפריטים שאינם `None`, והממוצע הוא עליהם:

## מה גודל ההגשה של כל מאמר?

הרשימה של המאמרים מראה את הכותרת, המחברים והנושאים של כל מאמר, אבל לא את הגודל של הקבצים שהמחברים הגישו. המספר הזה נמצא בדף של כל מאמר, שהרשימה מקשרת אליו. השאלה היא: בין שמונת המאמרים הראשונים ברשימה, לאיזה יש ההגשה הגדולה ביותר, ומה הגודל החציוני? כדי לענות, תוכנית צריכה לקרוא את הרשימה, לעקוב אחרי שמונה קישורים ולקרוא מספר אחד בכל דף.

התיבה הראשונה מגדירה את `soup_of`, פונקציה שמבקשת כתובת ומחזירה את ה־`soup` של הדף שלה, כמו ב[מדף אינטרנט לנתונים: Beautiful Soup](https://dvirross.com/he/courses/intro-data-science/data-collection-beautifulsoup), כדי שהדף הזה יעמוד בפני עצמו. היא בונה את ה־`soup` של הרשימה ואוספת את הקישורים למאמרים: כל קישור שהתכונה `title` שלו היא `Abstract`, כמו ב[מאלמנט אחד לכל הרשומות](https://dvirross.com/he/courses/intro-data-science/data-collection-find-all). חיתוך, `[:8]`, משאיר את שמונת הראשונים.

</div>

````python
import requests
from bs4 import BeautifulSoup

copies = (
    'https://dvirross.com/materials/courses/'
    'intro-data-science/'
)

def soup_of(address):
    text = requests.get(address).text
    return BeautifulSoup(text, 'html.parser')

soup = soup_of(copies + 'arxiv-cs-ar-2026-10.htm')
anchors = soup.find_all('a', title='Abstract')
links = [a.get('href') for a in anchors][:8]
print(links[0])
print(len(links))
````

## מהקישור אל הכתובת של דף

ה־`href` של מאמר הוא `/abs/2610.00106`: המזהה של המאמר הוא החלק האחרון שלו, והכתובת יחסית לאתר של הרשימה, `https://arxiv.org`. באתר החי הכתובת המלאה היא שם האתר ואחריו ה־`href`. העותקים השמורים של המאמרים נשמרים בשמות מהצורה `arxiv-abs-2610.00106.htm`, ולכן פונקציה הופכת `href` לכתובת של העותק שלו.

</div>

````python
def copy_address(href):
    paper_id = href.replace('/abs/', '')
    return copies + 'arxiv-abs-' + paper_id + '.htm'

page = soup_of(copy_address(links[0]))
print(page.find('h1', class_='title').text)
````

## מספר בתוך טקסט

הגודל של ההגשה כתוב בחלק של הדף שנקרא היסטוריית ההגשות, ב־`div` מהסוג `submission-history`, כחלק משורת טקסט: `[v1] Wed, 9 Sep 2026 02:40:29 UTC (3,330 KB)`. המספר נמצא בין סוגר לבין האותיות `KB`, ויש בו פסיק. מוציאים אותו בשלושה צעדים. `split('(')` חותכת את הטקסט בכל סוגר פותח ומחזירה רשימה של החלקים, ו־`[-1]`, החלק האחרון, הוא `'3,330 KB)'` עם מעבר שורה. לכל המאמרים האלה גרסה אחת; עם כמה גרסאות החלק האחרון הוא הגודל של האחרונה. `replace` מסירה את האותיות ואת הפסיק, `strip` מסירה את מעבר השורה, ו־`int` ממירה את הטקסט למספר. הפונקציה `repr` מראה טקסט כפי שהוא כתוב, עם התווים הנסתרים שבו, כמו מעבר השורה `\n`.

</div>

````python
box = page.find('div', class_='submission-history')
history = box.text
piece = history.split('(')[-1]
print(repr(piece))
number = piece.replace('KB)', '').replace(',', '')
kb = int(number.strip())
print(kb)
````

## אלמנט חסר עוצר את התוכנית

בחלק מהמאמרים יש בטבלה של הדף שורת הערות, אלמנט `td` מהסוג `comments`, ובחלקם אין. ההערה של המאמר הראשון קיימת; בדף של המאמר השני אין. לבקש את הטקסט של אלמנט שאינו שם זו שגיאה:

</div>

<div dir="rtl">

בדף השיעור התא שלמעלה נעצר בשגיאה בכוונה. במחברת התא שלמטה מדפיס את האלמנט ש־`find` החזירה, כדי שהמחברת תרוץ עד הסוף: הוא `None`, והשגיאה באה מכך שמבקשים מ־`None` את `.text`.

</div>

````python
second = soup_of(copy_address(links[1]))
print(second.find('td', class_='comments'))
````

<div dir="rtl">

`find` החזירה `None`, ול־`None` אין תכונה `text`. פונקציה שקוראת דף בודקת את האלמנט לפני שהיא קוראת ממנו, ומחזירה `None` בשביל הערך שלא קיים. הפונקציה שלמטה קוראת מאמר אחד ומחזירה את הנתונים שלו כמילון: המזהה, הכותרת מתגית ה־`meta`, הגודל ב־KB, וההערה או `None`.

</div>

````python
def read_paper(href):
    page = soup_of(copy_address(href))
    meta = {'name': 'citation_title'}
    title = page.find('meta', attrs=meta).get('content')
    box = page.find('div', class_='submission-history')
    piece = box.text.split('(')[-1]
    number = piece.replace('KB)', '').replace(',', '')
    kb = int(number.strip())
    comment = page.find('td', class_='comments')
    if comment is not None:
        comment = comment.text.strip()
    return {
        'id': href.replace('/abs/', ''),
        'title': title, 'kb': kb, 'comment': comment,
    }

second_paper = read_paper(links[1])
print(second_paper['kb'], second_paper['comment'])
````

## כל שמונת הדפים

לולאת `for`, או הבנת רשימה, קוראת לפונקציה על כל קישור. תוכנית ששולחת הרבה בקשות לאתר אחד ממתינה ביניהן, עם `time.sleep(seconds)`, כפי ש[לקרוא דפים באחריות](https://dvirross.com/he/courses/intro-data-science/data-collection-beautifulsoup#לקרוא-דפים-באחריות) מסביר; העותקים באתר הזה אינם arXiv, ולכן התיבה לא צריכה הפסקה.

</div>

````python
papers = [read_paper(href) for href in links]
for paper in papers:
    print(paper['id'], paper['kb'])
````

<div dir="rtl">

השאלה של השיעור נענית במקסימום, בממוצע ובחציון של הגדלים. החציון הוא האמצע של הגדלים לפי הסדר: `sorted` מסדרת אותם, ובשמונה ערכים הוא הממוצע של שני האמצעיים, במקומות 3 ו־4 (כשסופרים מ־0).

</div>

````python
sizes = [paper['kb'] for paper in papers]
largest = max(papers, key=lambda paper: paper['kb'])
print(largest['kb'], largest['title'][:40])
print(round(sum(sizes) / len(sizes)))
ordered = sorted(sizes)
print((ordered[3] + ordered[4]) / 2)
````

````python
import time
RUN_LIVE = False  # switch to True to ask arXiv itself (four requests, 15 seconds apart)
if RUN_LIVE:
    live_headers = {'User-Agent': 'IntroDataScienceCourse/1.0 (teaching example; contact via dvirross.com)'}
    live_list = requests.get('https://arxiv.org/list/cs.AR/2026-10?skip=0&show=25', headers=live_headers, timeout=30)
    refused = live_list.status_code in (403, 429)
    if not refused:
        live_soup = BeautifulSoup(live_list.text, 'html.parser')
        for link in live_soup.find_all('a', title='Abstract')[:3]:
            if not refused:
                time.sleep(15)
                live_page = requests.get('https://arxiv.org' + link.get('href'), headers=live_headers, timeout=30)
                print(live_page.status_code, link.get('href'))
                refused = live_page.status_code in (403, 429)
    if refused:
        print('The site refused a request or asked to slow down (status 403 or 429). Do not try again now; wait and use the saved copies.')
else:
    print('The live cell is off. Set RUN_LIVE = True to run it.')
````
